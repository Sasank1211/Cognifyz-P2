**Task 2: Rstuarant Recommendation System**

```python
# Step 1: Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Step 2: Load and preview the dataset

df = pd.read_csv("/content/Dataset .csv")

print("Dataset Preview:")
print(df.head())
print("\nColumns available:")
print(df.columns)

# Step 3: Handle missing values

df.fillna("", inplace=True)

# Step 4: Create a combined feature column

df['combined_features'] = (
    df['Cuisines'].astype(str) + " " +
    df['Price range'].astype(str) + " " +
    df['City'].astype(str)
)

# Step 5: Convert text data into numerical vectors

vectorizer = TfidfVectorizer(stop_words='english')
feature_vectors = vectorizer.fit_transform(df['combined_features'])

# Step 6: Compute similarity between all restaurants

similarity = cosine_similarity(feature_vectors)

# Step 7: Recommendation function

def recommend_restaurants(preferred_cuisine, preferred_city, preferred_price, top_n=5):
    """
    Recommends restaurants similar to user preferences.
    """
    # Create a user preference string similar to dataset format
    user_pref = f"{preferred_cuisine} {preferred_price} {preferred_city}"

    # Vectorize the user preference
    user_vector = vectorizer.transform([user_pref])

    # Compute similarity between user preference and all restaurants
    user_similarity = cosine_similarity(user_vector, feature_vectors)

    # Get top matches
    similar_indices = user_similarity.argsort()[0][-top_n:][::-1]

    # Return top recommended restaurants
    return df.iloc[similar_indices][['Restaurant Name', 'Cuisines', 'City', 'Price range', 'Aggregate rating']]

# Step 8: Test the recommendation system

sample_recommendations = recommend_restaurants(
    preferred_cuisine="Italian",
    preferred_city="New Delhi",
    preferred_price=3,
    top_n=5
)

print("\nTop 5 Restaurant Recommendations:")
print(sample_recommendations)
```
